```python
import pandas as pd

df = pd.read_csv("data/RYA22-24 subset250 data preproc v1.csv")

df.head()
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isnull().sum())
```

```python
print("Year:")
print(df["year"].value_counts(dropna=False))

print("\nSES:")
print(df["ses"].value_counts(dropna=False))

print("\nMissing year and SES together:")
print(df[df["year"].isna()]["ses"].value_counts(dropna=False))
```

```python
print("Missing SES by year:")
print(df[df["ses"].isna()]["year"].value_counts(dropna=False))

print("\nRows with missing year:")
display(df[df["year"].isna()].head(10))
```

```python
missing_year = df[df["year"].isna()]

print("Age range:", missing_year["age"].min(), "-", missing_year["age"].max())
print("Grade range:", missing_year["grade"].min(), "-", missing_year["grade"].max())

print("\nGender:")
print(missing_year["genderN"].value_counts())

print("\nCantril categories:")
print(missing_year["cantrilCat"].value_counts())
```

```python
# Check Cantril category consistency
def expected_cantril_category(score):
    if score <= 4:
        return "Struggling"
    elif score <= 6:
        return "Doing OK"
    else:
        return "Thriving"

df["expected_cantrilCat"] = df["cantril"].apply(expected_cantril_category)

cantril_errors = df[df["cantrilCat"] != df["expected_cantrilCat"]]

print("Cantril category errors:", len(cantril_errors))
display(cantril_errors.head())
```

```python
print("Age range:", df["age"].min(), "-", df["age"].max())
print("Grade range:", df["grade"].min(), "-", df["grade"].max())

print("\nGrade values:")
print(sorted(df["grade"].unique()))

print("\nGrade categories:")
print(df["gradeCatps"].value_counts().sort_index())

print("\nSchool categories:")
print(df["gradeCat"].value_counts())
```

```python
def expected_grade_group(grade):
    if grade in [3, 4]:
        return "Grades 3-4"
    elif grade in [5, 6]:
        return "Grades 5-6"
    elif grade in [7, 8]:
        return "Grades 7-8"
    elif grade in [9, 10]:
        return "Grades 9-10"
    elif grade in [11, 12]:
        return "Grades 11-12"

df["expected_gradeCatps"] = df["grade"].apply(expected_grade_group)

grade_group_errors = df[
    df["gradeCatps"] != df["expected_gradeCatps"]
]

print("Grade group errors:", len(grade_group_errors))
```

```python
df["expected_school"] = df["grade"].apply(
    lambda x: "primary school" if x <= 6 else "secondary school"
)

school_errors = df[
    df["gradeCat"] != df["expected_school"]
]

print("School category errors:", len(school_errors))
```

```python
print("Gender values:")
print(df["genderN"].value_counts(dropna=False))

print("\nYear values:")
print(df["year"].value_counts(dropna=False))

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
print("Cantril values:")
print(df["cantril"].value_counts().sort_index())

print("\nCantril range:")
print(df["cantril"].min(), "-", df["cantril"].max())
```
